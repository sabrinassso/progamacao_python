```python
import requests

# Classe para armazenar os dados climáticos
class CidadeClima:
    def __init__(self, nome, temperatura, umidade, condicao):
        self.nome = nome
        self.temperatura = temperatura
        self.umidade = umidade
        self.condicao = condicao

    def __str__(self):
        return (
            f"Cidade: {self.nome}\n"
            f"Temperatura: {self.temperatura}°C\n"
            f"Umidade: {self.umidade}%\n"
            f"Condição: {self.condicao}\n"
            + "-" * 40
        )


# Função para consultar o clima de uma cidade
def obter_clima(cidade, api_key):
    try:
        url = (
            f"https://api.openweathermap.org/data/2.5/weather"
            f"?q={cidade}&appid={api_key}&units=metric&lang=pt_br"
        )

        resposta = requests.get(url)

        if resposta.status_code == 200:
            dados = resposta.json()

            nome = dados["name"]
            temperatura = dados["main"]["temp"]
            umidade = dados["main"]["humidity"]
            condicao = dados["weather"][0]["description"]

            return CidadeClima(
                nome,
                temperatura,
                umidade,
                condicao
            )

        else:
            print(f"Erro ao obter dados de {cidade}.")
            return None

    except requests.exceptions.ConnectionError:
        print(f"Erro de conexão ao consultar {cidade}.")
        return None

    except Exception as erro:
        print(f"Erro inesperado em {cidade}: {erro}")
        return None


# Função para consultar várias cidades
def obter_clima_cidades(lista_cidades, api_key):
    relatorio_clima = []

    for cidade in lista_cidades:
        clima = obter_clima(cidade, api_key)

        if clima is not None:
            relatorio_clima.append(clima)

    return relatorio_clima


# Programa principal
if __name__ == "__main__":

    # Substitua pela sua chave da API
    api_key = "57f02dc11685a9611b42a5a22012380c"

    cidades = [
        "São Paulo",
        "Florianópolis",
        "Rio de Janeiro",
        "London",
        "Santa Maria"
    ]

    relatorio_clima = obter_clima_cidades(cidades, api_key)

    print("\nRELATÓRIO CLIMÁTICO")
    print("=" * 40)

    for cidade in relatorio_clima:
        print(cidade)
```
