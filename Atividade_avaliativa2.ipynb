```python
'''
Organizando dados
'''

with open ("/content/RPG.txt") as arquivo:
  conteudo = arquivo.read()

```

```python
lista_jogadores =  []
linhas =  conteudo.strip().split("\n")

for linha in linhas:
  dados = linha.split(";")

  if len(dados) == 5:
    jogador = {
    "nome":dados[0],
    "classe":dados[1],
    "kills":int(dados[2]),
    "deaths":int(dados[3]),
    "dano":int(dados[4])
  }

  lista_jogadores.append(jogador)


```

```python
def calcular_kda(kills, deaths):
  if deaths == 0:
    return kills
  return kills/deaths
```

```python
def filtrar_por_classse(lista, nome_classe):
  selecionados = []
  for j in lista:
    if j['classe'].lower() == nome_classe.lower():
      selecionados.append(j)
  return selecionados
```

```python
maior_dano = 0
nome_maior_dano = ""
soma_kills = 0

print("JOGADORES COM KDA SUPERIOR A 2.0:")
for j in lista_jogadores:
  kda = calcular_kda(j['kills'], j['deaths'])
  if kda > 2.0:
    print(f"- {j['nome'].upper()}")

    if j['dano'] > maior_dano:
      maior_dano = j['dano']
      nome_maior_dano = j['nome']

    soma_kills += j['kills']

  media_kills = soma_kills/len(lista_jogadores)
```

```python
print("-"*30)
print(f"MAIOR DANO: {nome_maior_dano} ({maior_dano})")
print(f"MEDIA DE KILLS: {media_kills}")
print("-"*30)
```

```python
magos = filtrar_por_classse(lista_jogadores, "Mago")
print(magos)

bruxa = filtrar_por_classse(lista_jogadores, "Bruxa")
print(bruxa)

guerreiro = filtrar_por_classse(lista_jogadores, "Guerreiro")
print(guerreiro)

arqueiro = filtrar_por_classse(lista_jogadores, "Arqueiro")
print(arqueiro)

vampiro = filtrar_por_classse(lista_jogadores, "Vampiro")
print(vampiro)
```
